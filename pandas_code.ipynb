## Data cleaning

```python
df = df.drop_duplicates()
```

# Cleaning names columns

```python
df['Name'] = df['Name'].str.strip().str.title()
```

```python
df.describe()
```

```python
df.loc[df['Billing Amount']<0, 'Billing Amount'] = df['Billing Amount'].mean()
```

```python
# Changing datatype
```

```python
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])
```

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
```

```python
df['Length_of_stay'] = (df['Discharge Date']-df['Date of Admission']).dt.days
```

## Data Analysis & Data Visualization

# 1)Which medical condition affects the most patients across the network?

```python
condition_count = df['Medical Condition'].value_counts()
condition_count
```

```python
plt.figure(figsize=(6,4))
plt.bar(condition_count.index,condition_count.values,color='#7833FE')
plt.title('Patients by medical condition')
plt.xlabel('Medical Condition')
plt.ylabel('Number of Patients')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

# 2) How is patient age distributed across the network?

```python
plt.figure(figsize=(6,4))
plt.hist(df['Age'], bins=20, color='red', edgecolor='black')
plt.title('Age distribution of patients')
plt.xlabel('Age')
plt.ylabel('Number of patients')
plt.show()
```

# 3)How has average treatment cost changed over time?

```python
df.head()
```

```python
avrg_cost = df['Billing Amount'].mean()
avrg_cost
```

```python
df['Admission Month'] = df['Date of Admission'].dt.to_period('M')
monthly_avrg_cost = df.groupby('Admission Month')['Billing Amount'].mean()
monthly_avrg_cost
```

```python
plt.figure(figsize=(11,4))
plt.plot(monthly_avrg_cost.index.astype(str), monthly_avrg_cost.values,color='#7833FE', marker='o',markersize=3)
plt.title('Average cost over time(Monthly)')
plt.xlabel('Month')
plt.ylabel('Average Billing Amount')
plt.xticks(rotation=90)
plt.show()
```

# 4) IS there any relationship between patient age and treatment cost ?

```python
senior_avg_cost = df.loc[df['Age']>= 60, 'Billing Amount'].mean()
senior_avg_cost
```

```python
sample_df = df.sample(1000, random_state=1)
plt.figure(figsize=(6,4))
plt.scatter(sample_df['Age'], sample_df['Billing Amount'], color ='purple', alpha=0.5, s=15)
plt.title('Age vs Treatment cost')
plt.xlabel('Age')
plt.ylabel('Billing Amount')
plt.show()
```

# 5) Which insurance providers cover the most patients?

```python
insurance_count = df['Insurance Provider'].value_counts()
insurance_count
```

```python
plt.figure(figsize=(5,5))
plt.pie(insurance_count.values,labels=insurance_count.index, autopct ='%1.1f%%', colors=['red', 'green', 'blue', 'purple', 'yellow'])
plt.title('Patient share by insurance provider')
plt.tight_layout()
plt.show()
```

# 6)How many emergency admissions occured in each month?

```python
df.head()
```

```python
emergency_adm = (df.loc[df['Admission Type']== 'Emergency'].groupby(df['Date of Admission'].dt.month).size())
emergency_adm 
```

# 7)Identify high cost patient groups.

```python
pd.options.display.float_format = '{:.2f}'.format
high_cost_patient = df.groupby('Medical Condition')['Billing Amount'].sum().sort_values(ascending=False)
high_cost_patient
```

# 8)Flagging emergency cases with abnormal results for review.

```python
priority_case = df.loc[(df['Admission Type']=='Emergency') & (df['Test Results']=='Abnormal')]
priority_case.shape[0]
priority_case
```

# 9)Grouping emergency_abnormal cases

```python
emergency_abnormal_cases = df.loc[(df['Admission Type']=='Emergency') & (df['Test Results']=='Abnormal'), ['Medical Condition']].groupby('Medical Condition').value_counts()
emergency_abnormal_cases.sort_values(ascending=False)
```
